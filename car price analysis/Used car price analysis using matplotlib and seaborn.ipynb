```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load dataset
df = pd.read_csv("Usedcarpriceanalysis.csv")
df.head()
```

```python
df.shape
```

```python
# fuel type distribution
sns.countplot(x='fuel',hue='fuel',data=df)
plt.title('Fuel Type Distribution')
plt.show()
```

#### Insights
##### Diesel cars dominate the listings with approximately 2200 entries
##### petrol cars follow,diesel.
##### very few listings for CNG/LPG
##### NO electric vehicle are listed in this dataset.
##### the absence of electric vehicle shows either very low adoption or that they are too new for resale.


```python
#price range analysis
sns.histplot(df['selling_price'],kde=True, bins=30, color='skyblue')
plt.title('price range analysis')
plt.xlabel('Price(INR)')
plt.show()
```

### Insights
##### Most cars falls under the 5lakh range
##### A long tail exist with premium cars priced 10 lakh range
##### the used car market is skewed(left skewed) towards budget buyers. a few luxary vehicle raise the maximum price
##### but the bulk of demand liesin the affordable segment

```python
# price by fueltype(boxplot)
sns.boxplot(x='fuel',hue='fuel',y='selling_price', data=df)
plt.title('price distribution by fueltype')            
plt.show()
```

### Insights
#### diesel vehicle have higher median selling price.petrol cars has higher spread but lower median.
#### diesel cars often hold better resale value due to fuel efficiency and higher initial price.CNG vehicle shows lower prices
#### often used for commercial or city travel.

```python
# transmission impact
sns.boxplot(x='transmission',hue='transmission',y='selling_price', data=df)
plt.title('selling price by transmission type')            
plt.show()
```

#### insights
#### Automatic cars shows higher median price.manual cars are more common with greater price variation.
#### Automatic transmission vehicle have higher resale value.possibly due to urban driving preference and convenience.

```python
#coreelation
correlation = df[['year','km_driven','selling_price']].corr()
sns.heatmap(correlation, annot=True, cmap='YlGnBu')
plt.title('Feature correlation heatmap')            
plt.show()
```

```python
###Insights
###selling price has a mild positive correlation with year.Mild negative correlation with km_driven.
###while no strong correlations,the age and usage of a car still show expected directional impact on resale value.
### No multicollinearity issue for basic regression models.
```
