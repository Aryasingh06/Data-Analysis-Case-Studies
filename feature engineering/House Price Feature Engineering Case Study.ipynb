## Feature engineering
Why Feature Engineering?

Raw features provide basic information.

For example:

Existing Feature - Limitation
PRICE  - Does not consider property size
BEDS - Does not indicate room efficiency
PROPERTYSQFT- Difficult to compare directly
LOCALITY - Text information only

To address these limitations, new features will be created.

```python
# Feature Creation
### Price Per Square Foot
df['PRICE_PER_SQFT'] = df['PRICE'] / df['PROPERTYSQFT']
```

```python
df[['PRICE','PROPERTYSQFT','PRICE_PER_SQFT']].head()
```

#### Two properties may have the same price but very different sizes. Price per square foot provides a more accurate measure of property value than total price alone

```python
# Total Rooms
df['TOTAL_ROOMS'] = df['BEDS'] + df['BATH']
```

```python
df[['BEDS','BATH',
'TOTAL_ROOMS']].head()
```

#### This feature provides a combined measure of room availability within a property.

```python
# Luxury Property Indicator
df['LUXURY_PROPERTY'] = (
df['PRICE'] > df['PRICE'].median()
).astype(int)
```

```python
df['LUXURY_PROPERTY'].value_counts()
```

### Creates a binary feature that distinguishes premium properties from standard properties.

```python
# Property Size Category
df['SIZE_CATEGORY'] = pd.cut(
df['PROPERTYSQFT'],
bins=[0,1000,2000,3000,10000],
labels=['Small','Medium','Large','Luxury']
)
```

```python
df['SIZE_CATEGORY'].value_counts()
```

#### Properties are segmented into meaningful size groups for easier analysis

```python
# Bedroom Density
df['BEDROOM_DENSITY'] = (
df['BEDS'] / df['PROPERTYSQFT']
)
```

```python
df[['BEDS',
'PROPERTYSQFT',
'BEDROOM_DENSITY']].head()
```

### Measures how efficiently available space is used

```python
# Bathroom Density
df['BATHROOM_DENSITY'] = (
df['BATH'] / df['PROPERTYSQFT']
)
```

```python
df[['BATH',
'PROPERTYSQFT',
'BATHROOM_DENSITY']].head()
```

### This feature measures bathroom availability relative to property size.

```python
# Log Price Transformation
import numpy as np

df['LOG_PRICE'] = np.log1p(df['PRICE'])
```

```python
df[['PRICE',
'LOG_PRICE']].head()
```

#### Log transformation reduces price skewness and stabilizes variance

```python
# Locality Frequency Encoding
locality_freq = df['LOCALITY'].value_counts()

df['LOCALITY_FREQ'] = (
df['LOCALITY'].map(locality_freq)
)
```

```python
df[['LOCALITY',
'LOCALITY_FREQ']].head()
```

#### This feature represents locality popularity based on listing frequency

```python
# Broker Property Count
broker_freq = (
df['BROKERTITLE'].value_counts()
)

df['BROKER_PROPERTY_COUNT'] = (
df['BROKERTITLE'].map(broker_freq)
)
```

```python
df[['BROKERTITLE',
'BROKER_PROPERTY_COUNT']].head()
```

#### This feature measures broker activity based on the number of listed properties

### Key Insights
- Property price alone does not fully represent a house's value; Price Per Square Foot provides a more meaningful   comparison.
- Combining bedrooms and bathrooms into Total Rooms gives a better view of property size and capacity.
- Categorizing properties by size helps simplify property segmentation and comparison.
- Frequency encoding transformed location information into numerical values while preserving locality popularity.
- Engineered features provide richer information than the original dataset and can support better analysis and machine learning models.

### Conclusion

Through feature engineering, raw property data was transformed into a feature-rich dataset. The newly created features provide better insights into house value, size, and location, making the dataset more useful for future analysis and machine learning applications
